```python
# Convert last_review to datetime 
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')

# Extract month from last_review
df['review_month'] = df['last_review'].dt.month

# Group by month and calculate average price
monthly_avg_price = df.groupby('review_month')['price'].mean()
```

```python
# Line Chart to see Average price by Month 
plt.figure(figsize=(10, 6))
sns.lineplot(x=monthly_avg_price.index, y=monthly_avg_price.values, marker='o', color='crimson')
plt.xticks(ticks=range(1, 13), labels=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.title('Average Price by Month', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Average Price ($)')
plt.grid(True)
plt.show()
```

**Insights -** This line chart illustrates the average price variation across different months. Prices peak in February, April, and July, while the lowest values are observed in January, August, and October. The trend suggests seasonal fluctuations in pricing

```python

plt.figure(figsize=(12, 6))
top_neighborhoods = df['neighbourhood'].value_counts().index[:10]  # Top 10 most listed neighborhoods

# Filter data for only top neighborhoods
df_filtered = df[df['neighbourhood'].isin(top_neighborhoods)]
```

```python
# Count Plot to see Room Type distribution across top 10 neighborhoods 
plt.figure(figsize=(14, 7)) 
sns.countplot(data=df_filtered, x='neighbourhood', hue='room_type', palette='coolwarm')
plt.xticks(rotation=45)
plt.title('Room Type Distribution Across Top 10 Neighborhoods', fontsize=14, fontweight='bold')
plt.xlabel('Neighborhood')
plt.ylabel('Count of Listings')
plt.legend(title='Room Type')
plt.show()
```

**Insights -** Most listings in the top 10 neighborhoods are Entire homes/apartments and Private rooms, with Harlem and Bedford-Stuyvesant having the highest private room counts. Hotel rooms and shared rooms are minimal, indicating Airbnb is primarily used for residential stays rather than traditional hospitality

```python
# Define availability categories
bins = [0, 90, 180, 270, 365]  # Availability ranges
labels = ['0-90 days', '91-180 days', '181-270 days', '271-365 days']
df['availability_category'] = pd.cut(df['availability_365'], bins=bins, labels=labels, include_lowest=True)

# Count of listings in each category for each room type
availability_counts = df.groupby(['room_type', 'availability_category']).size().unstack()
```

```python
# Stacked bar chart 
availability_counts.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='viridis')

plt.xlabel('Room Type', fontsize=14)
plt.ylabel('Number of Listings', fontsize=14)
plt.title('Availability Categories by Room Type', fontsize=14, fontweight='bold')
plt.xticks(rotation=0, fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title='Availability Range', fontsize=12, title_fontsize=14)

plt.show()
```

**Insights -** Most Airbnb listings are entire homes/apartments and private rooms, with a majority available for 271-365 days, indicating full-time rentals. Few listings are hotel or shared rooms, suggesting Airbnb is primarily used for residential stays. Some properties are available for 0-90 days, likely for seasonal rentals

```python
# Pie Chart for Room type Distributions 
room_type_counts = df['room_type'].value_counts()

# Pie chart 
plt.figure(figsize=(8, 8))
plt.pie(room_type_counts, labels=room_type_counts.index, autopct='%1.1f%%', colors=['royalblue', 'seagreen', 'tomato', 'gold'])
plt.title('Distribution of Room Types', fontsize=14, fontweight='bold')
plt.show()
```

**Insights -** The majority of listings are Entire home/apt (55.6%), followed by Private rooms (42.5%), indicating a strong preference for these options. Shared rooms (1.4%) and Hotel rooms (0.5%) make up a very small portion, showing limited demand for shared or hotel-like accommodations
